````python
import json
import os
import sys
from pathlib import Path

ROOT = Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))
from fhe_credit.experiments.executor import analyze_run, locate_run_results_root

RUN_ID = os.environ.get('HIST01_RUN_ID')
if RUN_ID:
    analysis = analyze_run(RUN_ID, results_root=locate_run_results_root(RUN_ID, ROOT))
    assert analysis['artifact_hashes_valid'], 'Artefatos ausentes ou adulterados.'
    print(json.dumps({'run_id': RUN_ID, 'status': analysis['status'], 'metrics': analysis['metrics']}, ensure_ascii=False, indent=2))
else:
    analysis = None
    print('Resultados não selecionados. Defina HIST01_RUN_ID e reexecute em modo analyze.')
````

## Resultados e visualizações

````python
if analysis is not None:
    import pandas as pd
    import matplotlib.pyplot as plt
    run_root = locate_run_results_root(RUN_ID, ROOT) / RUN_ID
    predictions = pd.read_csv(run_root / 'predictions.csv')
    predictions['score'].plot(kind='hist', bins=20, title='Distribuição dos escores — análise derivada')
    plt.xlabel('score')
    plt.show()
````
