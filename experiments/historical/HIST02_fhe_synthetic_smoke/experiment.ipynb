## Contexto e método

O script histórico usa 100 vetores sintéticos bidimensionais, árvore com `n_bits=3` e dois vetores de inferência. O relatório disponível compara rótulos nos modos quantizado em claro, simulado e FHE. Não é experimento de crédito nem benchmark.

````python
import json
import os
import sys
from pathlib import Path

ROOT = Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))
from fhe_credit.experiments.executor import analyze_run, locate_run_results_root

RUN_ID = os.environ.get('HIST02_RUN_ID')
if RUN_ID:
    analysis = analyze_run(RUN_ID, results_root=locate_run_results_root(RUN_ID, ROOT))
    assert analysis['artifact_hashes_valid'], 'Artefatos ausentes ou adulterados.'
    report = json.loads((locate_run_results_root(RUN_ID, ROOT) / RUN_ID / 'imported_report.json').read_text())
    print(json.dumps(report.get('comparison'), ensure_ascii=False, indent=2))
else:
    analysis = None
    print('Defina HIST02_RUN_ID para analisar uma importação verificável.')
````

## Resultados, proveniência e limitações

O relatório legado registra igualdade de rótulos nos três modos para dois vetores. O Run ID de importação registra o hash atual do JSON; esse hash não é histórico. Não há vetores, predições individuais, probabilidades, chaves, circuito, manifesto nem timestamp originais. Tempos disponíveis são medidas pontuais, não benchmarks robustos.
